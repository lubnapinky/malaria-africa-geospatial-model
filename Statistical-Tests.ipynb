```python
# ============================================================
# STANDALONE STATISTICAL COMPARISON ACROSS BURDEN CLUSTERS
# INCIDENCE MEAN ANALYSIS
#
# Input required:
#   1. df_Africa
#   2. cluster_0_countries
#   3. cluster_1_countries
#   4. cluster_2_countries
#
# What this code does:
#   - Creates cluster labels from country lists
#   - Computes country-level means for selected variables
#   - Runs Kruskal-Wallis tests
#   - Runs pairwise Mann-Whitney U tests with Holm correction
#   - Saves summary tables
#   - Makes publication-ready plots
#   - Uses consistent predictor-group colors in summary plots
# ============================================================

import os
import itertools
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import kruskal, mannwhitneyu

# ============================================================
# 0) USER INPUTS
# ============================================================
df_Africa = pd.read_csv('malaria-who-world-bank-Feb25-Africa.csv', encoding='latin1')

ANALYSIS_SUFFIX = "inc_mean"

cluster_0_countries = [
    "Zimbabwe", "Senegal", "Madagascar", "Somalia", "Mauritania",
    "Sudan", "Eritrea", "Namibia", "Djibouti", "Botswana",
    "South Africa", "eSwatini", "Algeria", "Morocco", "Egypt"
]

cluster_1_countries = [
    "Congo", "Zambia", "Angola", "Chad", "Gabon", "Gambia",
    "Rwanda", "Tanzania", "Guinea-Bissau", "Ethiopia", "Kenya"
]

cluster_2_countries = [
    "Burkina Faso", "Central African Rep.", "Benin", "Mali",
    "Dem. Rep. Congo", "Mozambique", "Sierra Leone", "Guinea",
    "Togo", "Côte d'Ivoire", "Uganda", "Niger", "Nigeria",
    "Malawi", "Cameroon", "Liberia", "Burundi", "Eq. Guinea",
    "Ghana", "S. Sudan"
]

CLUSTER_LABELS = {
    0: "Low burden",
    1: "Moderate burden",
    2: "High burden"
}

# ============================================================
# 1) SETTINGS
# ============================================================

OUTDIR_STATS = f"StatCom/{ANALYSIS_SUFFIX}"
os.makedirs(OUTDIR_STATS, exist_ok=True)

TITLE_FS = 18
LABEL_FS = 20
TICK_FS = 20
STAR_FS = 18

sns.set_style("white")

COLOR_MAP = {
    "Low burden": "#1b9e77",
    "Moderate burden": "#1f78b4",
    "High burden": "#d95f02",
}

ORDERED_LABELS = ["Low burden", "Moderate burden", "High burden"]

GROUP_COLOR_MAP = {
    "climate": "blue", #"#1f78b4",
    "land_environment": "#33a02c",
    "socioeconomic_demography": "purple", # "#6a3d9a",
    "wash": "cyan", #"#17becf",
    "health_system": "red", #"#e31a1c",
    "health_financing_coverage": "#ff7f00",
    "malaria_intervention": "pink", #"#b15928",
}


GROUP_LABEL_MAP = {
    "climate": "Climate",
    "land_environment": "Land Environment",
    "socioeconomic_demography": "Socioeconomic Demography",
    "wash": "WASH",
    "health_system": "Health system",
    "health_financing_coverage": "Health finance coverage",
    "malaria_intervention": "Malaria intervention"
}

# ============================================================
# 2) VARIABLE GROUPS
# ============================================================

VARIABLE_GROUPS = {

    "climate": [
        "Rainfall_mm_total",
        "Temp_C_avg",
        "Humidity_avg_pct",
    ],

    "land_environment": [
        "Agricultural_land",
        "Arable_land",
        "Forest_area",
        "Rural_land_area",
        "Permanent_cropland",
    ],

    "socioeconomic_demography": [
        "Access_to_electricity",
        "GDP_growth",
        "Literacy_rate",
        "Population_density",
        "Population_growth",
        "Population_total",
        "Rural_population",
    ],

    "wash": [
        "Mortality_rate_attributed_to_unsafe_WASH",
        "People_practicing_open_defecation",
        "People_using_at_least_basic_drinking_water_services",
        "People_using_at_least_basic_sanitation_services",
        "People_using_safely_managed_drinking_water_services",
        "People_using_safely_managed_sanitation_services",
        "People_with_basic_handwashing_facilities_including_soap_and_water",
    ],

    "health_system": [
        "Hospital_beds",
        "Nurses_and_midwives",
        "Physicians",
        "Community_health_workers",
    ],

    "health_financing_coverage": [
        "DGG_health_expenditure_per_capita",
        "UHC_service_coverage_sub-index_on_infectious_diseases",
    ],

    "malaria_intervention": [
        "Use_of_insecticide-treated_bed_nets",
    ]
}

# ============================================================
# 3) PRETTY NAMES
# ============================================================

PRETTY_NAMES = {
    "Rainfall_mm_total": "Total rainfall (mm)",
    "Temp_C_avg": "Average temperature (°C)",
    "Humidity_avg_pct": "Average relative humidity (%)",

    "Agricultural_land": "Agricultural land (% of land area)",
    "Arable_land": "Arable land (% of land area)",
    "Forest_area": "Forest area (% of land area)",
    "Rural_land_area": "Rural land area (sq. km)",
    "Permanent_cropland": "Permanent cropland (% of land area)",

    "GDP_growth": "GDP growth (annual %)",
    "Literacy_rate": "Adult literacy rate (%)",
    "Population_density": "Population density (people per sq. km)",
    "Population_growth": "Population growth (annual %)",
    "Population_total": "Total population",
    "Access_to_electricity": "Access to electricity (% of population)",
    "Rural_population": "Rural population (% of total population)",

    "People_practicing_open_defecation": "Population practicing open defecation (%)",
    "People_using_at_least_basic_drinking_water_services": "Access to basic drinking water services (%)",
    "People_using_at_least_basic_sanitation_services": "Access to basic sanitation services (%)",
    "People_using_safely_managed_drinking_water_services": "Access to safely managed drinking water (%)",
    "People_using_safely_managed_sanitation_services": "Access to safely managed sanitation (%)",
    "People_with_basic_handwashing_facilities_including_soap_and_water": "Access to basic handwashing facilities (%)",
    "Mortality_rate_attributed_to_unsafe_WASH": "Mortality rate attributed to unsafe WASH",

    "Hospital_beds": "Hospital beds (per 1,000 people)",
    "Physicians": "Physicians (per 1,000 people)",
    "Nurses_and_midwives": "Nurses and midwives (per 1,000 people)",
    "Community_health_workers": "Community health workers (per 1,000 people)",

    "DGG_health_expenditure_per_capita": "Government health expenditure per capita",
    "UHC_service_coverage_sub-index_on_infectious_diseases": "UHC infectious disease service coverage",

    "Use_of_insecticide-treated_bed_nets": "Use of insecticide-treated bed nets (%)"
}

# ============================================================
# 4) FILE LABELS
# ============================================================

FILE_LABELS = {
    "Rainfall_mm_total": "rainfall",
    "Temp_C_avg": "temp",
    "Humidity_avg_pct": "humidity",
    "Access_to_electricity": "electricity",
    "Agricultural_land": "agri_land",
    "Arable_land": "arable_land",
    "DGG_health_expenditure_per_capita": "gov_health_exp_pc",
    "Forest_area": "forest_area",
    "GDP_growth": "gdp_growth",
    "Hospital_beds": "hospital_beds",
    "Literacy_rate": "literacy",
    "Mortality_rate_attributed_to_unsafe_WASH": "unsafe_wash_mortality",
    "Nurses_and_midwives": "nurses_midwives",
    "Rural_land_area": "rural_land",
    "Rural_population": "rural_pop",
    "People_practicing_open_defecation": "open_defecation",
    "People_using_at_least_basic_drinking_water_services": "basic_drinking_water",
    "People_using_at_least_basic_sanitation_services": "basic_sanitation",
    "People_using_safely_managed_drinking_water_services": "safe_drinking_water",
    "People_using_safely_managed_sanitation_services": "safe_sanitation",
    "People_with_basic_handwashing_facilities_including_soap_and_water": "handwashing_basic",
    "Permanent_cropland": "permanent_cropland",
    "Physicians": "physicians",
    "Population_density": "pop_density",
    "Population_growth": "pop_growth",
    "Population_total": "pop_total",
    "UHC_service_coverage_sub-index_on_infectious_diseases": "uhc_infectious",
    "Use_of_insecticide-treated_bed_nets": "bed_nets",
    "Community_health_workers": "comm_health_workers"
}

# ============================================================
# 5) HELPER FUNCTIONS
# ============================================================

def clean_country_name(x):
    return str(x).strip()


def build_cluster_lookup():
    rows = []

    for cluster_id, countries in {
        0: cluster_0_countries,
        1: cluster_1_countries,
        2: cluster_2_countries
    }.items():

        for country in countries:
            rows.append({
                "Country": clean_country_name(country),
                "Cluster": cluster_id,
                "Cluster_Label": CLUSTER_LABELS[cluster_id]
            })

    cluster_df = pd.DataFrame(rows)

    duplicated = cluster_df[cluster_df.duplicated("Country", keep=False)]

    if not duplicated.empty:
        print("\nWARNING: Some countries appear in more than one cluster:")
        print(duplicated.sort_values("Country"))

    cluster_df = cluster_df.drop_duplicates("Country", keep="first")

    return cluster_df


def safe_filename(text, max_len=60):
    text = str(text)

    replacements = {
        "/": "_",
        "%": "pct",
        " ": "_",
        "(": "",
        ")": "",
        "-": "_",
        ",": "",
        "$": "usd"
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    while "__" in text:
        text = text.replace("__", "_")

    text = text.strip("_")

    if len(text) > max_len:
        text = text[:max_len].rstrip("_")

    return text


def get_significance_label(p):
    if p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    return None


def add_significance_brackets(ax, pairs, data, y_col, star_fs=14):
    if len(pairs) == 0:
        return

    y_max = data[y_col].max()
    y_min = data[y_col].min()
    y_range = y_max - y_min

    if pd.isna(y_range) or y_range == 0:
        y_range = max(abs(y_max), 1.0) * 0.2
        if y_range == 0:
            y_range = 1.0

    base_height = y_max + 0.08 * y_range
    step = 0.10 * y_range
    bracket_height = 0.03 * y_range

    for i, (x1, x2, pval) in enumerate(pairs):
        label = get_significance_label(pval)

        if label is None:
            continue

        y = base_height + i * step

        ax.plot(
            [x1, x1, x2, x2],
            [y, y + bracket_height, y + bracket_height, y],
            lw=1.4,
            c="black"
        )

        ax.text(
            (x1 + x2) / 2,
            y + bracket_height,
            label,
            ha="center",
            va="bottom",
            fontsize=star_fs
        )

    ax.set_ylim(top=base_height + len(pairs) * step + 0.12 * y_range)


def validate_inputs():
    if "df_Africa" not in globals():
        raise ValueError("df_Africa is not found. Load your dataset first.")

    if "Country" not in df_Africa.columns:
        raise ValueError("df_Africa must contain a 'Country' column.")


def build_country_level_analysis_data(variable_list, cluster_df):
    missing_vars = [v for v in variable_list if v not in df_Africa.columns]

    if missing_vars:
        raise ValueError(f"These variables are missing from df_Africa: {missing_vars}")

    temp = df_Africa[["Country"] + variable_list].copy()
    temp["Country"] = temp["Country"].astype(str).str.strip()

    for v in variable_list:
        temp[v] = pd.to_numeric(temp[v], errors="coerce")

    country_vars = (
        temp.groupby("Country", as_index=False)[variable_list]
        .mean()
    )

    analysis_df = cluster_df.merge(
        country_vars,
        on="Country",
        how="left"
    )

    missing_data_countries = analysis_df[
        analysis_df[variable_list].isna().all(axis=1)
    ]["Country"].tolist()

    if len(missing_data_countries) > 0:
        print("\nWARNING: These clustered countries were not found or have no data in df_Africa:")
        print(", ".join(missing_data_countries))

    return analysis_df


def format_sci(p):
    if p is None or np.isnan(p):
        return ""

    if p == 0:
        return r"$0$"

    exponent = int(np.floor(np.log10(p)))
    coeff = p / 10**exponent

    return rf"${coeff:.1f}\times 10^{{{exponent}}}$"


# ============================================================
# 6) GROUP-LEVEL SUMMARY PLOTS
# ============================================================

def plot_significant_features(sig_df, group_outdir, group_name):
    if sig_df.empty:
        print(f"No significant variables to plot for group: {group_name}")
        return

    plot_df = sig_df.copy().sort_values("p_value", ascending=True)
    plot_df["neg_log10_p"] = -np.log10(plot_df["p_value"].clip(lower=1e-300))

    plt.figure(figsize=(9, max(4, 0.45 * len(plot_df))))

    ax = sns.barplot(
        data=plot_df,
        y="Pretty_name",
        x="neg_log10_p",
        color="gray"
    )

    ax.set_xlabel(r"$-\log_{10}(p)$", fontsize=LABEL_FS)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=TICK_FS)
    ax.tick_params(axis="y", labelsize=TICK_FS)
    ax.grid(False)

    plt.tight_layout()

    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_p.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_p.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    plot_df2 = sig_df.copy().sort_values("epsilon_squared", ascending=False)

    plt.figure(figsize=(9, max(4, 0.45 * len(plot_df2))))

    ax = sns.barplot(
        data=plot_df2,
        y="Pretty_name",
        x="epsilon_squared",
        color="gray"
    )

    ax.set_xlabel("Epsilon-squared", fontsize=LABEL_FS)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=TICK_FS)
    ax.tick_params(axis="y", labelsize=TICK_FS)
    ax.grid(False)

    plt.tight_layout()

    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_ES.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_ES.pdf"),
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 7) OVERALL SUMMARY PLOTS WITH CONSISTENT COLORS
# ============================================================

def plot_all_summaries_consistent_colors(
    all_sig_df,
    outdir,
    title_fs=16,
    axis_fs=16,
    tick_fs=14,
    label_fs=13,
    legend_fs=12,
    show_titles=False
):

    if all_sig_df is None or all_sig_df.empty:
        print("No significant variables available for plotting.")
        return

    os.makedirs(outdir, exist_ok=True)

    plot_df = all_sig_df.copy()

    plot_df["p_value"] = np.asarray(plot_df["p_value"], dtype=float)
    plot_df["epsilon_squared"] = np.asarray(plot_df["epsilon_squared"], dtype=float)

    plot_df["neg_log10_p"] = -np.log10(plot_df["p_value"].clip(lower=1e-300))
    plot_df["Group_Display"] = plot_df["Group"].map(GROUP_LABEL_MAP).fillna(plot_df["Group"])

    display_color_map = {
        GROUP_LABEL_MAP.get(k, k): v for k, v in GROUP_COLOR_MAP.items()
    }

    groups_present = plot_df["Group_Display"].dropna().unique().tolist()

    palette_used = {
        g: display_color_map.get(g, "#7f7f7f")
        for g in groups_present
    }

    plot_df_p = plot_df.sort_values("neg_log10_p", ascending=True).reset_index(drop=True)
    plot_df_e = plot_df.sort_values("epsilon_squared", ascending=True).reset_index(drop=True)

    # --------------------------------------------------------
    # 1) Summary by p-value
    # --------------------------------------------------------

    plt.figure(figsize=(16, max(5, 0.55 * len(plot_df_p))))

    ax = sns.barplot(
        data=plot_df_p,
        y="Pretty_name",
        x="neg_log10_p",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90
    )

    x_max = plot_df_p["neg_log10_p"].max()
    ax.set_xlim(0, x_max * 1.25)

    for i, (_, row) in enumerate(plot_df_p.iterrows()):
        ax.text(
            row["neg_log10_p"] + (x_max * 0.015),
            i,
            format_sci(row["p_value"]),
            va="center",
            ha="left",
            fontsize=label_fs
        )

    ax.axvline(-np.log10(0.05), linestyle="--", linewidth=1, color="black")
    ax.axvline(-np.log10(0.01), linestyle=":", linewidth=1, color="black")

    if show_titles:
        ax.set_title("Statistical significance by predictor group", fontsize=title_fs)

    ax.set_xlabel(r"$-\log_{10}(p\text{-value})$", fontsize=axis_fs)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=tick_fs)
    ax.tick_params(axis="y", labelsize=tick_fs)
    ax.grid(False)

    ax.legend(
        title="Predictor group",
        loc="upper right",
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_pvalue.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_pvalue.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # 2) Summary by effect size
    # --------------------------------------------------------

    plt.figure(figsize=(16, max(5, 0.55 * len(plot_df_e))))

    ax = sns.barplot(
        data=plot_df_e,
        y="Pretty_name",
        x="epsilon_squared",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90
    )

    x_max = plot_df_e["epsilon_squared"].max()
    ax.set_xlim(0, x_max * 1.20)

    for i, (_, row) in enumerate(plot_df_e.iterrows()):
        ax.text(
            row["epsilon_squared"] + (x_max * 0.015),
            i,
            f"{row['epsilon_squared']:.2f}",
            va="center",
            ha="left",
            fontsize=label_fs
        )

    if show_titles:
        ax.set_title("Effect size by predictor group", fontsize=title_fs)

    ax.set_xlabel(r"Effect size ($\epsilon^2$)", fontsize=axis_fs)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=tick_fs)
    ax.tick_params(axis="y", labelsize=tick_fs)
    ax.grid(False)

    ax.legend(
        title="Predictor group",
        loc="upper right",
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_effect_size.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_effect_size.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # 3) Dual-panel summary
    # --------------------------------------------------------

    plot_df_dual = plot_df_p.copy()

    fig, axes = plt.subplots(
        ncols=2,
        figsize=(20, max(5, 0.55 * len(plot_df_dual))),
        sharey=True
    )

    ax1, ax2 = axes

    sns.barplot(
        data=plot_df_dual,
        y="Pretty_name",
        x="neg_log10_p",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90,
        ax=ax1
    )

    x_max_p = plot_df_dual["neg_log10_p"].max()
    ax1.set_xlim(0, x_max_p * 1.25)

    for i, (_, row) in enumerate(plot_df_dual.iterrows()):
        ax1.text(
            row["neg_log10_p"] + (x_max_p * 0.015),
            i,
            format_sci(row["p_value"]),
            va="center",
            ha="left",
            fontsize=label_fs
        )

    ax1.axvline(-np.log10(0.05), linestyle="--", linewidth=1, color="black")
    ax1.axvline(-np.log10(0.01), linestyle=":", linewidth=1, color="black")

    if show_titles:
        ax1.set_title("Significance", fontsize=title_fs)

    ax1.set_xlabel(r"$-\log_{10}(p)$", fontsize=axis_fs)
    ax1.set_ylabel("")
    ax1.tick_params(axis="x", labelsize=tick_fs)
    ax1.tick_params(axis="y", labelsize=tick_fs)
    ax1.grid(False)

    sns.barplot(
        data=plot_df_dual,
        y="Pretty_name",
        x="epsilon_squared",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90,
        ax=ax2
    )

    x_max_e = plot_df_dual["epsilon_squared"].max()
    ax2.set_xlim(0, x_max_e * 1.25)

    for i, (_, row) in enumerate(plot_df_dual.iterrows()):
        ax2.text(
            row["epsilon_squared"] + (x_max_e * 0.015),
            i,
            f"{row['epsilon_squared']:.2f}",
            va="center",
            ha="left",
            fontsize=label_fs
        )

    if show_titles:
        ax2.set_title("Effect size", fontsize=title_fs)

    ax2.set_xlabel(r"Effect size ($\epsilon^2$)", fontsize=axis_fs)
    ax2.set_ylabel("")
    ax2.tick_params(axis="x", labelsize=tick_fs)
    ax2.tick_params(axis="y", labelsize=tick_fs)
    ax2.grid(False)

    handles, labels = ax1.get_legend_handles_labels()

    if ax1.legend_ is not None:
        ax1.legend_.remove()

    if ax2.legend_ is not None:
        ax2.legend_.remove()

    fig.legend(
        handles,
        labels,
        title="Predictor group",
        loc="lower center",
        bbox_to_anchor=(0.5, -0.12),
        ncol=3,
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout(rect=[0, 0.06, 1, 1])

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_dual.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_dual.pdf"),
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 8) MAIN GROUP COMPARISON FUNCTION
# ============================================================

def run_comparison_for_group(group_name, variable_list, cluster_df):

    group_outdir = os.path.join(OUTDIR_STATS, group_name)
    os.makedirs(group_outdir, exist_ok=True)

    print("\n" + "=" * 80)
    print(f"Running group: {group_name}")
    print("=" * 80)

    analysis_df = build_country_level_analysis_data(variable_list, cluster_df)

    analysis_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_country_level_analysis_data.csv"),
        index=False
    )

    with open(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_countries_by_burden_cluster.txt"),
        "w",
        encoding="utf-8"
    ) as f:

        for c in sorted(analysis_df["Cluster"].dropna().unique()):
            label = CLUSTER_LABELS[c]
            countries = (
                analysis_df.loc[analysis_df["Cluster"] == c, "Country"]
                .sort_values()
                .tolist()
            )

            print(f"\nCluster {c} ({label}) – {len(countries)} countries:")
            print(", ".join(countries))

            f.write(f"Cluster {c} ({label}) – {len(countries)} countries:\n")
            f.write(", ".join(countries) + "\n\n")

    summary_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        for c in sorted(sub["Cluster"].unique()):
            vals = sub.loc[sub["Cluster"] == c, var]
            label = CLUSTER_LABELS[c]

            summary_rows.append({
                "Variable": var,
                "Pretty_name": PRETTY_NAMES.get(var, var),
                "Cluster": c,
                "Cluster_Label": label,
                "n": vals.shape[0],
                "mean": vals.mean(),
                "median": vals.median(),
                "sd": vals.std(),
                "min": vals.min(),
                "max": vals.max()
            })

    summary_df = pd.DataFrame(summary_rows)

    summary_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_clstrws_sum.csv"),
        index=False
    )

    kw_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        groups = [
            sub.loc[sub["Cluster"] == c, var].values
            for c in sorted(sub["Cluster"].unique())
        ]

        if len(groups) < 2:
            continue

        H, p = kruskal(*groups)

        n = len(sub)
        k_groups = len(groups)
        epsilon_sq = (H - k_groups + 1) / (n - k_groups) if n > k_groups else np.nan

        kw_rows.append({
            "Variable": var,
            "Pretty_name": PRETTY_NAMES.get(var, var),
            "N_countries_used": n,
            "Kruskal_H": H,
            "p_value": p,
            "epsilon_squared": epsilon_sq
        })

    kw_df = pd.DataFrame(kw_rows)

    if not kw_df.empty:
        kw_df = kw_df.sort_values("p_value")

    kw_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_krskl.csv"),
        index=False
    )

    sig_kw_df = kw_df[kw_df["p_value"] < 0.05].copy()

    if not sig_kw_df.empty:
        sig_kw_df["Group"] = group_name

        sig_kw_df.to_csv(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_signfcnt_vars.csv"),
            index=False
        )

        print("\nSignificant variables:")
        print(sig_kw_df[["Pretty_name", "Kruskal_H", "p_value", "epsilon_squared"]])

        plot_significant_features(sig_kw_df, group_outdir, group_name)

    else:
        print("\nNo significant variables at p < 0.05.")

    pw_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()
        raw_tests = []

        for a, b in itertools.combinations(sorted(sub["Cluster"].unique()), 2):
            x = sub.loc[sub["Cluster"] == a, var]
            y = sub.loc[sub["Cluster"] == b, var]

            if len(x) == 0 or len(y) == 0:
                continue

            U, p = mannwhitneyu(x, y, alternative="two-sided")

            raw_tests.append({
                "Variable": var,
                "Pretty_name": PRETTY_NAMES.get(var, var),
                "Cluster_A": a,
                "Label_A": CLUSTER_LABELS[a],
                "Cluster_B": b,
                "Label_B": CLUSTER_LABELS[b],
                "n_A": len(x),
                "n_B": len(y),
                "U_stat": U,
                "raw_p": p
            })

        raw_tests = sorted(raw_tests, key=lambda d: d["raw_p"])
        m = len(raw_tests)

        for i, d in enumerate(raw_tests):
            d["holm_p"] = min((m - i) * d["raw_p"], 1.0)
            d["significant_holm_0.05"] = d["holm_p"] < 0.05
            pw_rows.append(d)

    pw_df = pd.DataFrame(pw_rows)

    pw_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_pairwse_posthoc_MW.csv"),
        index=False
    )

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        if sub.empty:
            continue

        sub["Cluster_Label"] = pd.Categorical(
            sub["Cluster_Label"],
            categories=ORDERED_LABELS,
            ordered=True
        )

        sub = sub.sort_values("Cluster_Label")

        pretty_label = PRETTY_NAMES.get(var, var)
        fname = FILE_LABELS.get(var, safe_filename(var))

        sig_pairs = []
        sub_pw = pw_df[pw_df["Variable"] == var].copy()

        for _, row in sub_pw.iterrows():
            if bool(row["significant_holm_0.05"]):
                x1 = ORDERED_LABELS.index(row["Label_A"])
                x2 = ORDERED_LABELS.index(row["Label_B"])
                sig_pairs.append((x1, x2, float(row["holm_p"])))

        sig_pairs = sorted(sig_pairs, key=lambda x: (x[0], x[1]))

        plt.figure(figsize=(8.5, 6))

        ax = sns.boxplot(
            x="Cluster_Label",
            y=var,
            data=sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            legend=False
        )

        sns.stripplot(
            x="Cluster_Label",
            y=var,
            data=sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            size=4,
            alpha=0.70,
            legend=False
        )

        add_significance_brackets(ax, sig_pairs, sub, var, star_fs=STAR_FS)

        ax.set_xlabel("Burden cluster", fontsize=LABEL_FS)
        ax.set_ylabel(pretty_label, fontsize=LABEL_FS)
        ax.tick_params(axis="x", labelsize=TICK_FS)
        ax.tick_params(axis="y", labelsize=TICK_FS)
        ax.grid(False)

        plt.tight_layout()

        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_boxplot_sig.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_boxplot_sig.pdf"),
            bbox_inches="tight"
        )

        plt.show()

        mean_sub = (
            sub.groupby("Cluster_Label", as_index=False, observed=False)[var]
            .mean()
        )

        mean_sub["Cluster_Label"] = pd.Categorical(
            mean_sub["Cluster_Label"],
            categories=ORDERED_LABELS,
            ordered=True
        )

        mean_sub = mean_sub.sort_values("Cluster_Label")

        plt.figure(figsize=(7.5, 5.5))

        ax = sns.barplot(
            x="Cluster_Label",
            y=var,
            data=mean_sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            legend=False
        )

        ax.set_xlabel("Burden cluster", fontsize=LABEL_FS)
        ax.set_ylabel(pretty_label, fontsize=LABEL_FS)
        ax.tick_params(axis="x", labelsize=TICK_FS)
        ax.tick_params(axis="y", labelsize=TICK_FS)
        ax.grid(False)

        plt.tight_layout()

        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_barplot.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_barplot.pdf"),
            bbox_inches="tight"
        )

        plt.show()

    return summary_df, kw_df, pw_df, sig_kw_df


# ============================================================
# 9) RUN PIPELINE
# ============================================================

validate_inputs()

cluster_df = build_cluster_lookup()

cluster_df.to_csv(
    os.path.join(OUTDIR_STATS, f"{ANALYSIS_SUFFIX}_cntry_clstr_assgnmnts.csv"),
    index=False
)

all_results = {}
all_sig_rows = []

for group_name, variable_list in VARIABLE_GROUPS.items():

    if len(variable_list) == 0:
        continue

    summary_df, kw_df, pw_df, sig_kw_df = run_comparison_for_group(
        group_name=group_name,
        variable_list=variable_list,
        cluster_df=cluster_df
    )

    all_results[group_name] = {
        "summary": summary_df,
        "kruskal": kw_df,
        "pairwise": pw_df,
        "significant": sig_kw_df
    }

    if sig_kw_df is not None and not sig_kw_df.empty:
        all_sig_rows.append(sig_kw_df)


# ============================================================
# 10) OVERALL SIGNIFICANT VARIABLE SUMMARY
# ============================================================

if len(all_sig_rows) > 0:

    all_sig_df = pd.concat(all_sig_rows, ignore_index=True)

    all_sig_df = all_sig_df.sort_values(
        ["p_value", "epsilon_squared"],
        ascending=[True, False]
    )

    all_sig_df.to_csv(
        os.path.join(OUTDIR_STATS, f"{ANALYSIS_SUFFIX}_all_grps_signfcnt_vars.csv"),
        index=False
    )

    print("\nSIGNIFICANT VARIABLE NAMES ONLY")
    print("=" * 80)
    
    significant_variable_names = all_sig_df["Variable"].drop_duplicates().tolist()
    
    for var in significant_variable_names:
        print(var)

    print("\n" + "=" * 80)
    print("OVERALL SIGNIFICANT VARIABLES")
    print("=" * 80)
    
    print(
        all_sig_df[
            ["Group", "Pretty_name", "Kruskal_H", "p_value", "epsilon_squared"]
        ]
    )

    plot_all_summaries_consistent_colors(
        all_sig_df=all_sig_df,
        outdir=OUTDIR_STATS,
        title_fs=16,
        axis_fs=20,
        tick_fs=20,
        label_fs=20,
        legend_fs=18,
        show_titles=False
    )

else:
    all_sig_df = pd.DataFrame()
    print("\nNo significant variables found across any groups.")

print(f"\nDone. Outputs saved in: {OUTDIR_STATS}")
```

```python
# ============================================================
# STANDALONE STATISTICAL COMPARISON ACROSS BURDEN CLUSTERS
# INCIDENCE MEAN ANALYSIS
#
# Input required:
#   1. df_Africa
#   2. cluster_0_countries
#   3. cluster_1_countries
#   4. cluster_2_countries
#
# What this code does:
#   - Creates cluster labels from country lists
#   - Computes country-level means for selected variables
#   - Runs Kruskal-Wallis tests
#   - Runs pairwise Mann-Whitney U tests with Holm correction
#   - Saves summary tables
#   - Makes publication-ready plots
#   - Uses consistent predictor-group colors in summary plots
# ============================================================

import os
import itertools
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import kruskal, mannwhitneyu

# ============================================================
# 0) USER INPUTS
# ============================================================
df_Africa = pd.read_csv('malaria-who-world-bank-Feb25-Africa.csv', encoding='latin1')

ANALYSIS_SUFFIX = "mort_mean"

# ============================================================
# MORTALITY MEAN BURDEN CLUSTERS
# (2000–2024 average mortality burden)
# ============================================================

# ------------------------------------------------------------
# Cluster 0 — Low burden
# ------------------------------------------------------------

cluster_0_countries = [
    "Rwanda",
    "Gambia",
    "Namibia",
    "Ethiopia",
    "Kenya",
    "Zimbabwe",
    "Madagascar",
    "Gabon",
    "Mauritania",
    "Senegal",
    "Somalia",
    "Sudan",
    "Eritrea",
    "eSwatini",
    "Djibouti",
    "South Africa",
    "Botswana",
    "Algeria",
    "Egypt",
    "Morocco"
]

# ------------------------------------------------------------
# Cluster 1 — Moderate burden
# ------------------------------------------------------------

cluster_1_countries = [
    "Benin",
    "Liberia",
    "Côte d'Ivoire",
    "Chad",
    "S. Sudan",
    "Uganda",
    "Burundi",
    "Cameroon",
    "Malawi",
    "Angola",
    "Togo",
    "Guinea-Bissau",
    "Zambia",
    "Tanzania",
    "Ghana",
    "Eq. Guinea",
    "Congo"
]

# ------------------------------------------------------------
# Cluster 2 — High burden
# ------------------------------------------------------------

cluster_2_countries = [
    "Burkina Faso",
    "Sierra Leone",
    "Niger",
    "Dem. Rep. Congo",
    "Mali",
    "Central African Rep.",
    "Nigeria",
    "Mozambique",
    "Guinea"
]


CLUSTER_LABELS = {
    0: "Low burden",
    1: "Moderate burden",
    2: "High burden"
}

# ============================================================
# 1) SETTINGS
# ============================================================

OUTDIR_STATS = f"StatCom/{ANALYSIS_SUFFIX}"
os.makedirs(OUTDIR_STATS, exist_ok=True)

TITLE_FS = 18
LABEL_FS = 20
TICK_FS = 20
STAR_FS = 18

sns.set_style("white")

COLOR_MAP = {
    "Low burden": "#1b9e77",
    "Moderate burden": "#1f78b4",
    "High burden": "#d95f02",
}

ORDERED_LABELS = ["Low burden", "Moderate burden", "High burden"]

GROUP_COLOR_MAP = {
    "climate":  "blue", #"#1f78b4",
    "land_environment": "#33a02c",
    "socioeconomic_demography": "purple", # "#6a3d9a",
    "wash": "cyan", #"#17becf",
    "health_system": "red", #"#e31a1c",
    "health_financing_coverage": "#ff7f00",
    "malaria_intervention": "pink", #"#b15928",
}


GROUP_LABEL_MAP = {
    "climate": "Climate",
    "land_environment": "Land Environment",
    "socioeconomic_demography": "Socioeconomic Demography",
    "wash": "WASH",
    "health_system": "Health system",
    "health_financing_coverage": "Health finance coverage",
    "malaria_intervention": "Malaria intervention"
}

# ============================================================
# 2) VARIABLE GROUPS
# ============================================================

VARIABLE_GROUPS = {

    "climate": [
        "Rainfall_mm_total",
        "Temp_C_avg",
        "Humidity_avg_pct",
    ],

    "land_environment": [
        "Agricultural_land",
        "Arable_land",
        "Forest_area",
        "Rural_land_area",
        "Permanent_cropland",
    ],

    "socioeconomic_demography": [
        "Access_to_electricity",
        "GDP_growth",
        "Literacy_rate",
        "Population_density",
        "Population_growth",
        "Population_total",
        "Rural_population",
    ],

    "wash": [
        "Mortality_rate_attributed_to_unsafe_WASH",
        "People_practicing_open_defecation",
        "People_using_at_least_basic_drinking_water_services",
        "People_using_at_least_basic_sanitation_services",
        "People_using_safely_managed_drinking_water_services",
        "People_using_safely_managed_sanitation_services",
        "People_with_basic_handwashing_facilities_including_soap_and_water",
    ],

    "health_system": [
        "Hospital_beds",
        "Nurses_and_midwives",
        "Physicians",
        "Community_health_workers",
    ],

    "health_financing_coverage": [
        "DGG_health_expenditure_per_capita",
        "UHC_service_coverage_sub-index_on_infectious_diseases",
    ],

    "malaria_intervention": [
        "Use_of_insecticide-treated_bed_nets",
    ]
}

# ============================================================
# 3) PRETTY NAMES
# ============================================================

PRETTY_NAMES = {
    "Rainfall_mm_total": "Total rainfall (mm)",
    "Temp_C_avg": "Average temperature (°C)",
    "Humidity_avg_pct": "Average relative humidity (%)",

    "Agricultural_land": "Agricultural land (% of land area)",
    "Arable_land": "Arable land (% of land area)",
    "Forest_area": "Forest area (% of land area)",
    "Rural_land_area": "Rural land area (sq. km)",
    "Permanent_cropland": "Permanent cropland (% of land area)",

    "GDP_growth": "GDP growth (annual %)",
    "Literacy_rate": "Adult literacy rate (%)",
    "Population_density": "Population density (people per sq. km)",
    "Population_growth": "Population growth (annual %)",
    "Population_total": "Total population",
    "Access_to_electricity": "Access to electricity (% of population)",
    "Rural_population": "Rural population (% of total population)",

    "People_practicing_open_defecation": "Population practicing open defecation (%)",
    "People_using_at_least_basic_drinking_water_services": "Access to basic drinking water services (%)",
    "People_using_at_least_basic_sanitation_services": "Access to basic sanitation services (%)",
    "People_using_safely_managed_drinking_water_services": "Access to safely managed drinking water (%)",
    "People_using_safely_managed_sanitation_services": "Access to safely managed sanitation (%)",
    "People_with_basic_handwashing_facilities_including_soap_and_water": "Access to basic handwashing facilities (%)",
    "Mortality_rate_attributed_to_unsafe_WASH": "Mortality rate attributed to unsafe WASH",

    "Hospital_beds": "Hospital beds (per 1,000 people)",
    "Physicians": "Physicians (per 1,000 people)",
    "Nurses_and_midwives": "Nurses and midwives (per 1,000 people)",
    "Community_health_workers": "Community health workers (per 1,000 people)",

    "DGG_health_expenditure_per_capita": "Government health expenditure per capita",
    "UHC_service_coverage_sub-index_on_infectious_diseases": "UHC infectious disease service coverage",

    "Use_of_insecticide-treated_bed_nets": "Use of insecticide-treated bed nets (%)"
}

# ============================================================
# 4) FILE LABELS
# ============================================================

FILE_LABELS = {
    "Rainfall_mm_total": "rainfall",
    "Temp_C_avg": "temp",
    "Humidity_avg_pct": "humidity",
    "Access_to_electricity": "electricity",
    "Agricultural_land": "agri_land",
    "Arable_land": "arable_land",
    "DGG_health_expenditure_per_capita": "gov_health_exp_pc",
    "Forest_area": "forest_area",
    "GDP_growth": "gdp_growth",
    "Hospital_beds": "hospital_beds",
    "Literacy_rate": "literacy",
    "Mortality_rate_attributed_to_unsafe_WASH": "unsafe_wash_mortality",
    "Nurses_and_midwives": "nurses_midwives",
    "Rural_land_area": "rural_land",
    "Rural_population": "rural_pop",
    "People_practicing_open_defecation": "open_defecation",
    "People_using_at_least_basic_drinking_water_services": "basic_drinking_water",
    "People_using_at_least_basic_sanitation_services": "basic_sanitation",
    "People_using_safely_managed_drinking_water_services": "safe_drinking_water",
    "People_using_safely_managed_sanitation_services": "safe_sanitation",
    "People_with_basic_handwashing_facilities_including_soap_and_water": "handwashing_basic",
    "Permanent_cropland": "permanent_cropland",
    "Physicians": "physicians",
    "Population_density": "pop_density",
    "Population_growth": "pop_growth",
    "Population_total": "pop_total",
    "UHC_service_coverage_sub-index_on_infectious_diseases": "uhc_infectious",
    "Use_of_insecticide-treated_bed_nets": "bed_nets",
    "Community_health_workers": "comm_health_workers"
}

# ============================================================
# 5) HELPER FUNCTIONS
# ============================================================

def clean_country_name(x):
    return str(x).strip()


def build_cluster_lookup():
    rows = []

    for cluster_id, countries in {
        0: cluster_0_countries,
        1: cluster_1_countries,
        2: cluster_2_countries
    }.items():

        for country in countries:
            rows.append({
                "Country": clean_country_name(country),
                "Cluster": cluster_id,
                "Cluster_Label": CLUSTER_LABELS[cluster_id]
            })

    cluster_df = pd.DataFrame(rows)

    duplicated = cluster_df[cluster_df.duplicated("Country", keep=False)]

    if not duplicated.empty:
        print("\nWARNING: Some countries appear in more than one cluster:")
        print(duplicated.sort_values("Country"))

    cluster_df = cluster_df.drop_duplicates("Country", keep="first")

    return cluster_df


def safe_filename(text, max_len=60):
    text = str(text)

    replacements = {
        "/": "_",
        "%": "pct",
        " ": "_",
        "(": "",
        ")": "",
        "-": "_",
        ",": "",
        "$": "usd"
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    while "__" in text:
        text = text.replace("__", "_")

    text = text.strip("_")

    if len(text) > max_len:
        text = text[:max_len].rstrip("_")

    return text


def get_significance_label(p):
    if p < 0.001:
        return "***"
    elif p < 0.01:
        return "**"
    elif p < 0.05:
        return "*"
    return None


def add_significance_brackets(ax, pairs, data, y_col, star_fs=14):
    if len(pairs) == 0:
        return

    y_max = data[y_col].max()
    y_min = data[y_col].min()
    y_range = y_max - y_min

    if pd.isna(y_range) or y_range == 0:
        y_range = max(abs(y_max), 1.0) * 0.2
        if y_range == 0:
            y_range = 1.0

    base_height = y_max + 0.08 * y_range
    step = 0.10 * y_range
    bracket_height = 0.03 * y_range

    for i, (x1, x2, pval) in enumerate(pairs):
        label = get_significance_label(pval)

        if label is None:
            continue

        y = base_height + i * step

        ax.plot(
            [x1, x1, x2, x2],
            [y, y + bracket_height, y + bracket_height, y],
            lw=1.4,
            c="black"
        )

        ax.text(
            (x1 + x2) / 2,
            y + bracket_height,
            label,
            ha="center",
            va="bottom",
            fontsize=star_fs
        )

    ax.set_ylim(top=base_height + len(pairs) * step + 0.12 * y_range)


def validate_inputs():
    if "df_Africa" not in globals():
        raise ValueError("df_Africa is not found. Load your dataset first.")

    if "Country" not in df_Africa.columns:
        raise ValueError("df_Africa must contain a 'Country' column.")


def build_country_level_analysis_data(variable_list, cluster_df):
    missing_vars = [v for v in variable_list if v not in df_Africa.columns]

    if missing_vars:
        raise ValueError(f"These variables are missing from df_Africa: {missing_vars}")

    temp = df_Africa[["Country"] + variable_list].copy()
    temp["Country"] = temp["Country"].astype(str).str.strip()

    for v in variable_list:
        temp[v] = pd.to_numeric(temp[v], errors="coerce")

    country_vars = (
        temp.groupby("Country", as_index=False)[variable_list]
        .mean()
    )

    analysis_df = cluster_df.merge(
        country_vars,
        on="Country",
        how="left"
    )

    missing_data_countries = analysis_df[
        analysis_df[variable_list].isna().all(axis=1)
    ]["Country"].tolist()

    if len(missing_data_countries) > 0:
        print("\nWARNING: These clustered countries were not found or have no data in df_Africa:")
        print(", ".join(missing_data_countries))

    return analysis_df


def format_sci(p):
    if p is None or np.isnan(p):
        return ""

    if p == 0:
        return r"$0$"

    exponent = int(np.floor(np.log10(p)))
    coeff = p / 10**exponent

    return rf"${coeff:.1f}\times 10^{{{exponent}}}$"


# ============================================================
# 6) GROUP-LEVEL SUMMARY PLOTS
# ============================================================

def plot_significant_features(sig_df, group_outdir, group_name):
    if sig_df.empty:
        print(f"No significant variables to plot for group: {group_name}")
        return

    plot_df = sig_df.copy().sort_values("p_value", ascending=True)
    plot_df["neg_log10_p"] = -np.log10(plot_df["p_value"].clip(lower=1e-300))

    plt.figure(figsize=(9, max(4, 0.45 * len(plot_df))))

    ax = sns.barplot(
        data=plot_df,
        y="Pretty_name",
        x="neg_log10_p",
        color="gray"
    )

    ax.set_xlabel(r"$-\log_{10}(p)$", fontsize=LABEL_FS)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=TICK_FS)
    ax.tick_params(axis="y", labelsize=TICK_FS)
    ax.grid(False)

    plt.tight_layout()

    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_p.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_p.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    plot_df2 = sig_df.copy().sort_values("epsilon_squared", ascending=False)

    plt.figure(figsize=(9, max(4, 0.45 * len(plot_df2))))

    ax = sns.barplot(
        data=plot_df2,
        y="Pretty_name",
        x="epsilon_squared",
        color="gray"
    )

    ax.set_xlabel("Epsilon-squared", fontsize=LABEL_FS)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=TICK_FS)
    ax.tick_params(axis="y", labelsize=TICK_FS)
    ax.grid(False)

    plt.tight_layout()

    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_ES.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_ES.pdf"),
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 7) OVERALL SUMMARY PLOTS WITH CONSISTENT COLORS
# ============================================================

def plot_all_summaries_consistent_colors(
    all_sig_df,
    outdir,
    title_fs=16,
    axis_fs=16,
    tick_fs=14,
    label_fs=13,
    legend_fs=12,
    show_titles=False
):

    if all_sig_df is None or all_sig_df.empty:
        print("No significant variables available for plotting.")
        return

    os.makedirs(outdir, exist_ok=True)

    plot_df = all_sig_df.copy()

    plot_df["p_value"] = np.asarray(plot_df["p_value"], dtype=float)
    plot_df["epsilon_squared"] = np.asarray(plot_df["epsilon_squared"], dtype=float)

    plot_df["neg_log10_p"] = -np.log10(plot_df["p_value"].clip(lower=1e-300))
    plot_df["Group_Display"] = plot_df["Group"].map(GROUP_LABEL_MAP).fillna(plot_df["Group"])

    display_color_map = {
        GROUP_LABEL_MAP.get(k, k): v for k, v in GROUP_COLOR_MAP.items()
    }

    groups_present = plot_df["Group_Display"].dropna().unique().tolist()

    palette_used = {
        g: display_color_map.get(g, "#7f7f7f")
        for g in groups_present
    }

    plot_df_p = plot_df.sort_values("neg_log10_p", ascending=True).reset_index(drop=True)
    plot_df_e = plot_df.sort_values("epsilon_squared", ascending=True).reset_index(drop=True)

    # --------------------------------------------------------
    # 1) Summary by p-value
    # --------------------------------------------------------

    plt.figure(figsize=(16, max(5, 0.55 * len(plot_df_p))))

    ax = sns.barplot(
        data=plot_df_p,
        y="Pretty_name",
        x="neg_log10_p",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90
    )

    x_max = plot_df_p["neg_log10_p"].max()
    ax.set_xlim(0, x_max * 1.25)

    for i, (_, row) in enumerate(plot_df_p.iterrows()):
        ax.text(
            row["neg_log10_p"] + (x_max * 0.015),
            i,
            format_sci(row["p_value"]),
            va="center",
            ha="left",
            fontsize=label_fs
        )

    ax.axvline(-np.log10(0.05), linestyle="--", linewidth=1, color="black")
    ax.axvline(-np.log10(0.01), linestyle=":", linewidth=1, color="black")

    if show_titles:
        ax.set_title("Statistical significance by predictor group", fontsize=title_fs)

    ax.set_xlabel(r"$-\log_{10}(p\text{-value})$", fontsize=axis_fs)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=tick_fs)
    ax.tick_params(axis="y", labelsize=tick_fs)
    ax.grid(False)

    ax.legend(
        title="Predictor group",
        loc="upper right",
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_pvalue.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_pvalue.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # 2) Summary by effect size
    # --------------------------------------------------------

    plt.figure(figsize=(16, max(5, 0.55 * len(plot_df_e))))

    ax = sns.barplot(
        data=plot_df_e,
        y="Pretty_name",
        x="epsilon_squared",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90
    )

    x_max = plot_df_e["epsilon_squared"].max()
    ax.set_xlim(0, x_max * 1.20)

    for i, (_, row) in enumerate(plot_df_e.iterrows()):
        ax.text(
            row["epsilon_squared"] + (x_max * 0.015),
            i,
            f"{row['epsilon_squared']:.2f}",
            va="center",
            ha="left",
            fontsize=label_fs
        )

    if show_titles:
        ax.set_title("Effect size by predictor group", fontsize=title_fs)

    ax.set_xlabel(r"Effect size ($\epsilon^2$)", fontsize=axis_fs)
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=tick_fs)
    ax.tick_params(axis="y", labelsize=tick_fs)
    ax.grid(False)

    ax.legend(
        title="Predictor group",
        loc="upper right",
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_effect_size.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_effect_size.pdf"),
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # 3) Dual-panel summary
    # --------------------------------------------------------

    plot_df_dual = plot_df_p.copy()

    fig, axes = plt.subplots(
        ncols=2,
        figsize=(20, max(5, 0.55 * len(plot_df_dual))),
        sharey=True
    )

    ax1, ax2 = axes

    sns.barplot(
        data=plot_df_dual,
        y="Pretty_name",
        x="neg_log10_p",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90,
        ax=ax1
    )

    x_max_p = plot_df_dual["neg_log10_p"].max()
    ax1.set_xlim(0, x_max_p * 1.25)

    for i, (_, row) in enumerate(plot_df_dual.iterrows()):
        ax1.text(
            row["neg_log10_p"] + (x_max_p * 0.015),
            i,
            format_sci(row["p_value"]),
            va="center",
            ha="left",
            fontsize=label_fs
        )

    ax1.axvline(-np.log10(0.05), linestyle="--", linewidth=1, color="black")
    ax1.axvline(-np.log10(0.01), linestyle=":", linewidth=1, color="black")

    if show_titles:
        ax1.set_title("Significance", fontsize=title_fs)

    ax1.set_xlabel(r"$-\log_{10}(p)$", fontsize=axis_fs)
    ax1.set_ylabel("")
    ax1.tick_params(axis="x", labelsize=tick_fs)
    ax1.tick_params(axis="y", labelsize=tick_fs)
    ax1.grid(False)

    sns.barplot(
        data=plot_df_dual,
        y="Pretty_name",
        x="epsilon_squared",
        hue="Group_Display",
        palette=palette_used,
        dodge=False,
        edgecolor="black",
        alpha=0.90,
        ax=ax2
    )

    x_max_e = plot_df_dual["epsilon_squared"].max()
    ax2.set_xlim(0, x_max_e * 1.25)

    for i, (_, row) in enumerate(plot_df_dual.iterrows()):
        ax2.text(
            row["epsilon_squared"] + (x_max_e * 0.015),
            i,
            f"{row['epsilon_squared']:.2f}",
            va="center",
            ha="left",
            fontsize=label_fs
        )

    if show_titles:
        ax2.set_title("Effect size", fontsize=title_fs)

    ax2.set_xlabel(r"Effect size ($\epsilon^2$)", fontsize=axis_fs)
    ax2.set_ylabel("")
    ax2.tick_params(axis="x", labelsize=tick_fs)
    ax2.tick_params(axis="y", labelsize=tick_fs)
    ax2.grid(False)

    handles, labels = ax1.get_legend_handles_labels()

    if ax1.legend_ is not None:
        ax1.legend_.remove()

    if ax2.legend_ is not None:
        ax2.legend_.remove()

    fig.legend(
        handles,
        labels,
        title="Predictor group",
        loc="lower center",
        bbox_to_anchor=(0.5, -0.15),
        ncol=3,
        fontsize=legend_fs,
        title_fontsize=legend_fs,
        frameon=True
    )

    plt.tight_layout(rect=[0, 0.06, 1, 1])

    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_dual.png"),
        dpi=300,
        bbox_inches="tight"
    )
    plt.savefig(
        os.path.join(outdir, f"{ANALYSIS_SUFFIX}_summary_dual.pdf"),
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 8) MAIN GROUP COMPARISON FUNCTION
# ============================================================

def run_comparison_for_group(group_name, variable_list, cluster_df):

    group_outdir = os.path.join(OUTDIR_STATS, group_name)
    os.makedirs(group_outdir, exist_ok=True)

    print("\n" + "=" * 80)
    print(f"Running group: {group_name}")
    print("=" * 80)

    analysis_df = build_country_level_analysis_data(variable_list, cluster_df)

    analysis_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_country_level_analysis_data.csv"),
        index=False
    )

    with open(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_countries_by_burden_cluster.txt"),
        "w",
        encoding="utf-8"
    ) as f:

        for c in sorted(analysis_df["Cluster"].dropna().unique()):
            label = CLUSTER_LABELS[c]
            countries = (
                analysis_df.loc[analysis_df["Cluster"] == c, "Country"]
                .sort_values()
                .tolist()
            )

            print(f"\nCluster {c} ({label}) – {len(countries)} countries:")
            print(", ".join(countries))

            f.write(f"Cluster {c} ({label}) – {len(countries)} countries:\n")
            f.write(", ".join(countries) + "\n\n")

    summary_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        for c in sorted(sub["Cluster"].unique()):
            vals = sub.loc[sub["Cluster"] == c, var]
            label = CLUSTER_LABELS[c]

            summary_rows.append({
                "Variable": var,
                "Pretty_name": PRETTY_NAMES.get(var, var),
                "Cluster": c,
                "Cluster_Label": label,
                "n": vals.shape[0],
                "mean": vals.mean(),
                "median": vals.median(),
                "sd": vals.std(),
                "min": vals.min(),
                "max": vals.max()
            })

    summary_df = pd.DataFrame(summary_rows)

    summary_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_clstrws_sum.csv"),
        index=False
    )

    kw_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        groups = [
            sub.loc[sub["Cluster"] == c, var].values
            for c in sorted(sub["Cluster"].unique())
        ]

        if len(groups) < 2:
            continue

        H, p = kruskal(*groups)

        n = len(sub)
        k_groups = len(groups)
        epsilon_sq = (H - k_groups + 1) / (n - k_groups) if n > k_groups else np.nan

        kw_rows.append({
            "Variable": var,
            "Pretty_name": PRETTY_NAMES.get(var, var),
            "N_countries_used": n,
            "Kruskal_H": H,
            "p_value": p,
            "epsilon_squared": epsilon_sq
        })

    kw_df = pd.DataFrame(kw_rows)

    if not kw_df.empty:
        kw_df = kw_df.sort_values("p_value")

    kw_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_krskl.csv"),
        index=False
    )

    sig_kw_df = kw_df[kw_df["p_value"] < 0.05].copy()

    if not sig_kw_df.empty:
        sig_kw_df["Group"] = group_name

        sig_kw_df.to_csv(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_signfcnt_vars.csv"),
            index=False
        )

        print("\nSignificant variables:")
        print(sig_kw_df[["Pretty_name", "Kruskal_H", "p_value", "epsilon_squared"]])

        plot_significant_features(sig_kw_df, group_outdir, group_name)

    else:
        print("\nNo significant variables at p < 0.05.")

    pw_rows = []

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()
        raw_tests = []

        for a, b in itertools.combinations(sorted(sub["Cluster"].unique()), 2):
            x = sub.loc[sub["Cluster"] == a, var]
            y = sub.loc[sub["Cluster"] == b, var]

            if len(x) == 0 or len(y) == 0:
                continue

            U, p = mannwhitneyu(x, y, alternative="two-sided")

            raw_tests.append({
                "Variable": var,
                "Pretty_name": PRETTY_NAMES.get(var, var),
                "Cluster_A": a,
                "Label_A": CLUSTER_LABELS[a],
                "Cluster_B": b,
                "Label_B": CLUSTER_LABELS[b],
                "n_A": len(x),
                "n_B": len(y),
                "U_stat": U,
                "raw_p": p
            })

        raw_tests = sorted(raw_tests, key=lambda d: d["raw_p"])
        m = len(raw_tests)

        for i, d in enumerate(raw_tests):
            d["holm_p"] = min((m - i) * d["raw_p"], 1.0)
            d["significant_holm_0.05"] = d["holm_p"] < 0.05
            pw_rows.append(d)

    pw_df = pd.DataFrame(pw_rows)

    pw_df.to_csv(
        os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{group_name}_pairwse_posthoc_MW.csv"),
        index=False
    )

    for var in variable_list:
        sub = analysis_df[["Country", "Cluster", "Cluster_Label", var]].dropna().copy()

        if sub.empty:
            continue

        sub["Cluster_Label"] = pd.Categorical(
            sub["Cluster_Label"],
            categories=ORDERED_LABELS,
            ordered=True
        )

        sub = sub.sort_values("Cluster_Label")

        pretty_label = PRETTY_NAMES.get(var, var)
        fname = FILE_LABELS.get(var, safe_filename(var))

        sig_pairs = []
        sub_pw = pw_df[pw_df["Variable"] == var].copy()

        for _, row in sub_pw.iterrows():
            if bool(row["significant_holm_0.05"]):
                x1 = ORDERED_LABELS.index(row["Label_A"])
                x2 = ORDERED_LABELS.index(row["Label_B"])
                sig_pairs.append((x1, x2, float(row["holm_p"])))

        sig_pairs = sorted(sig_pairs, key=lambda x: (x[0], x[1]))

        plt.figure(figsize=(8.5, 6))

        ax = sns.boxplot(
            x="Cluster_Label",
            y=var,
            data=sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            legend=False
        )

        sns.stripplot(
            x="Cluster_Label",
            y=var,
            data=sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            size=4,
            alpha=0.70,
            legend=False
        )

        add_significance_brackets(ax, sig_pairs, sub, var, star_fs=STAR_FS)

        ax.set_xlabel("Burden cluster", fontsize=LABEL_FS)
        ax.set_ylabel(pretty_label, fontsize=LABEL_FS)
        ax.tick_params(axis="x", labelsize=TICK_FS)
        ax.tick_params(axis="y", labelsize=TICK_FS)
        ax.grid(False)

        plt.tight_layout()

        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_boxplot_sig.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_boxplot_sig.pdf"),
            bbox_inches="tight"
        )

        plt.show()

        mean_sub = (
            sub.groupby("Cluster_Label", as_index=False, observed=False)[var]
            .mean()
        )

        mean_sub["Cluster_Label"] = pd.Categorical(
            mean_sub["Cluster_Label"],
            categories=ORDERED_LABELS,
            ordered=True
        )

        mean_sub = mean_sub.sort_values("Cluster_Label")

        plt.figure(figsize=(7.5, 5.5))

        ax = sns.barplot(
            x="Cluster_Label",
            y=var,
            data=mean_sub,
            hue="Cluster_Label",
            palette=COLOR_MAP,
            dodge=False,
            legend=False
        )

        ax.set_xlabel("Burden cluster", fontsize=LABEL_FS)
        ax.set_ylabel(pretty_label, fontsize=LABEL_FS)
        ax.tick_params(axis="x", labelsize=TICK_FS)
        ax.tick_params(axis="y", labelsize=TICK_FS)
        ax.grid(False)

        plt.tight_layout()

        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_barplot.png"),
            dpi=300,
            bbox_inches="tight"
        )
        plt.savefig(
            os.path.join(group_outdir, f"{ANALYSIS_SUFFIX}_{fname}_barplot.pdf"),
            bbox_inches="tight"
        )

        plt.show()

    return summary_df, kw_df, pw_df, sig_kw_df


# ============================================================
# 9) RUN PIPELINE
# ============================================================

validate_inputs()

cluster_df = build_cluster_lookup()

cluster_df.to_csv(
    os.path.join(OUTDIR_STATS, f"{ANALYSIS_SUFFIX}_cntry_clstr_assgnmnts.csv"),
    index=False
)

all_results = {}
all_sig_rows = []

for group_name, variable_list in VARIABLE_GROUPS.items():

    if len(variable_list) == 0:
        continue

    summary_df, kw_df, pw_df, sig_kw_df = run_comparison_for_group(
        group_name=group_name,
        variable_list=variable_list,
        cluster_df=cluster_df
    )

    all_results[group_name] = {
        "summary": summary_df,
        "kruskal": kw_df,
        "pairwise": pw_df,
        "significant": sig_kw_df
    }

    if sig_kw_df is not None and not sig_kw_df.empty:
        all_sig_rows.append(sig_kw_df)


# ============================================================
# 10) OVERALL SIGNIFICANT VARIABLE SUMMARY
# ============================================================

if len(all_sig_rows) > 0:

    all_sig_df = pd.concat(all_sig_rows, ignore_index=True)

    all_sig_df = all_sig_df.sort_values(
        ["p_value", "epsilon_squared"],
        ascending=[True, False]
    )

    all_sig_df.to_csv(
        os.path.join(OUTDIR_STATS, f"{ANALYSIS_SUFFIX}_all_grps_signfcnt_vars.csv"),
        index=False
    )

    print("\nSIGNIFICANT VARIABLE NAMES ONLY")
    print("=" * 80)
    
    significant_variable_names = all_sig_df["Variable"].drop_duplicates().tolist()
    
    for var in significant_variable_names:
        print(var)

    print("\n" + "=" * 80)
    print("OVERALL SIGNIFICANT VARIABLES")
    print("=" * 80)

    print(
        all_sig_df[
            ["Group", "Pretty_name", "Kruskal_H", "p_value", "epsilon_squared"]
        ]
    )

    plot_all_summaries_consistent_colors(
        all_sig_df=all_sig_df,
        outdir=OUTDIR_STATS,
        title_fs=16,
        axis_fs=20,
        tick_fs=20,
        label_fs=20,
        legend_fs=18,
        show_titles=False
    )

else:
    all_sig_df = pd.DataFrame()
    print("\nNo significant variables found across any groups.")

print(f"\nDone. Outputs saved in: {OUTDIR_STATS}")
```
